This project focuses on emotion classification for Twitter posts. I experimented with two transformer-based models (BERTweet-Base and BERTweet-Large) (Nguyen, Vu & Nguyen, 2020) and one LLM embedding and XGBoost model using Ollama embeddings. In the next sections I describe how preprocessing, feature engineering, and modeling were handled.

I used BERTweet for classifying tweets because previous work has achieved state-of-the-art performance on this task by combining techniques such as bidirectional GRUs, variational autoencoders, word and document embeddings, self-attention, graph attention and adversarial training and  BERTweet, a variant of BERT pre-trained specifically on a Twitter corpus, with its transformer architecture achieved competitive performance with the existing literature (Albada & Sonola, 2025). 

On the other hand, I used a combination of Ollama text embedding model (snowflake-arctic-embed) with XGBoost because it relates to the techniques seen in the laboratory homework. I also picked this particular Ollama embedding model because at the time of their release, the model achieved state-of-the-art retrieval accuracy (Merrick, Xu, Nuti & Campos, 2024).

### 1.3 Ollama Embeddings + XGBoost

## 2. Feature Engineering Steps

### 2.1 BERTweet-Base

```python
# Model & tokenizer
model_name = "vinai/bertweet-base"
tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=False)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=len(labels))

# Datasets
train_dataset = TweetDataset(df_train_, tokenizer)
val_dataset = TweetDataset(df_val_, tokenizer)
test_dataset = TweetDataset(df_test, tokenizer)
```

```python
# Model & tokenizer
model_name = "vinai/bertweet-large"
tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=False)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=len(labels))

# Datasets
train_dataset = TweetDataset(df_train_, tokenizer)
val_dataset = TweetDataset(df_val_, tokenizer)
test_dataset = TweetDataset(df_test, tokenizer)
```

### 2.3 Ollama Embeddings + XGBoost

```python
# Embeddings
def embed_text(text, model="snowflake-arctic-embed"):
    resp = ollama.embeddings(model=model, prompt=text)
    return resp["embedding"]

def embed_texts(texts, model="snowflake-arctic-embed", max_workers=8):
    vectors = []
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = {executor.submit(embed_text, t, model): t for t in texts}
        for f in tqdm(as_completed(futures), total=len(texts), desc="Embedding texts"):
            vectors.append(f.result())
    return np.array(vectors)

print("▶ Generating TRAIN embeddings...")
train_emb = embed_texts(X_train)

print("▶ Generating VAL embeddings...")
val_emb = embed_texts(X_val)

print("▶ Generating TEST embeddings...")
test_emb = embed_texts(df_test["text"].tolist())
```

## 3. Model Implementation Steps

### 3.1 BERTweet-Base

```python
# Metric
def compute_metrics(p):
    preds = p.predictions.argmax(axis=1)
    return {"f1": f1_score(p.label_ids, preds, average="macro")}

# Training
training_args = TrainingArguments(
    output_dir="./bertweet_emotion",
    num_train_epochs=3,
    per_device_train_batch_size=32,
    learning_rate=5e-5,
    weight_decay=0.01,
    logging_steps=50,
    eval_steps=200,
    save_strategy="steps",
    save_steps=200,
    metric_for_best_model="f1",
    greater_is_better=True,
    fp16=True
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics
)

trainer.train()

# Predict test
preds = trainer.predict(test_dataset).predictions
pred_labels = [id2label[i] for i in preds.argmax(axis=1)]

# Save
pd.DataFrame({"id": df_test["id"].tolist(), "emotion": pred_labels}).to_csv("BERTweet_base_2.csv", index=False)
print("Saved BERTweet_base_2.csv")
```

### 3.2 BERTweet-Large

```python
# Metric
def compute_metrics(p):
    preds = p.predictions.argmax(axis=1)
    return {"f1": f1_score(p.label_ids, preds, average="macro")}

# Training
training_args = TrainingArguments(
    output_dir="./bertweet_large_emotion",
    num_train_epochs=3,                  # max epochs, early stopping will cut off if no improvement
    per_device_train_batch_size=32,
    gradient_accumulation_steps=4,
    learning_rate=5e-5,
    weight_decay=0.01,
    logging_steps=50,
    eval_steps=200,
    save_strategy="steps",
    save_steps=200,
    metric_for_best_model="f1",
    greater_is_better=True,
    fp16=True,
    save_total_limit=2,
    disable_tqdm=False
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics,
    # callbacks=[EarlyStoppingCallback(early_stopping_patience=3)]  # stops if val F1 doesn't improve for 3 evals
)

trainer.train()

# Predict test
preds = trainer.predict(test_dataset).predictions
pred_labels = [id2label[i] for i in preds.argmax(axis=1)]

# Save
pd.DataFrame({"id": df_test["id"].tolist(), "emotion": pred_labels}).to_csv("BERTweet_large_1.csv", index=False)
print("Saved BERTweet_large_1.csv")
```

### 3.3 Ollama Embeddings + XGBoost

```python
# Training
clf = xgb.XGBClassifier(
    n_estimators=700,
    max_depth=8,
    learning_rate=0.05,
    subsample=0.9,
    colsample_bytree=0.8,
    objective="multi:softprob",
    eval_metric="mlogloss",
    tree_method="hist",           # GPU-compatible
    predictor="gpu_predictor"
)

clf.fit(train_emb, y_train)

# Validate
val_pred = clf.predict(val_emb)
f1 = f1_score(y_val, val_pred, average="macro")
print(f"\nVALIDATION MACRO F1: {f1:.4f}\n")

# Predict test
test_pred_ids = clf.predict(test_emb)
test_pred_labels = label_enc.inverse_transform(test_pred_ids)

# Save
df_out = pd.DataFrame({
    "id": df_test["id"].tolist(),
    "emotion": test_pred_labels
})

df_out.to_csv("ollama_1.csv", index=False)
print("Saved to ollama_1.csv")
```
